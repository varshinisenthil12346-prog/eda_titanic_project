```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
os.makedirs("output", exist_ok=True)

# 1. LOAD DATA
df = sns.load_dataset("titanic")
print("Shape:", df.shape)
print("\nColumns:", list(df.columns))
print("\nData types:\n", df.dtypes)
print("\nFirst 5 rows:\n", df.head())

# 2. MISSING VALUES + CLEANING
print("\nMissing values:\n", df.isnull().sum()[df.isnull().sum() > 0])
df = df.drop(columns=["deck"])                       # too many missing
df["age"] = df["age"].fillna(df["age"].median())
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode()[0])
print("Missing after cleaning:", int(df.isnull().sum().sum()))

# 3. STATISTICAL SUMMARY
print("\n===== NUMERIC SUMMARY =====")
print(df.describe().round(2))
print("\n===== CATEGORICAL SUMMARY =====")
print(df.describe(include=["object", "category"]))
df.describe().round(2).to_csv("output/summary_statistics.csv")

# 4. UNIVARIATE ANALYSIS
fig, ax = plt.subplots(2, 2, figsize=(13, 9))
sns.histplot(df["age"], bins=30, kde=True, ax=ax[0, 0], color="steelblue")
ax[0, 0].set_title("Age Distribution")
sns.histplot(df["fare"], bins=30, kde=True, ax=ax[0, 1], color="darkorange")
ax[0, 1].set_title("Fare Distribution")
sns.countplot(data=df, x="pclass", hue="pclass", legend=False, ax=ax[1, 0])
ax[1, 0].set_title("Passengers by Class")
sns.countplot(data=df, x="survived", hue="survived", legend=False, ax=ax[1, 1])
ax[1, 1].set_title("Survived (0 = No, 1 = Yes)")
plt.suptitle("Univariate Analysis", fontsize=16, fontweight="bold")
plt.tight_layout()
plt.savefig("output/1_univariate.png", dpi=150)
plt.show()

# 5. BIVARIATE ANALYSIS (what affects survival?)
df["age_group"] = pd.cut(df["age"], bins=[0, 12, 18, 35, 60, 100],
                         labels=["Child", "Teen", "Young Adult", "Adult", "Senior"])

fig, ax = plt.subplots(2, 3, figsize=(18, 10))
sns.barplot(data=df, x="sex", y="survived", hue="sex", errorbar=None, legend=False, ax=ax[0, 0])
ax[0, 0].set_title("Survival Rate by Gender")
sns.barplot(data=df, x="pclass", y="survived", hue="pclass", errorbar=None, legend=False, ax=ax[0, 1])
ax[0, 1].set_title("Survival Rate by Class")
sns.barplot(data=df, x="age_group", y="survived", hue="age_group", errorbar=None, legend=False, ax=ax[0, 2])
ax[0, 2].set_title("Survival Rate by Age Group")
sns.barplot(data=df, x="embarked", y="survived", hue="embarked", errorbar=None, legend=False, ax=ax[1, 0])
ax[1, 0].set_title("Survival Rate by Port")
sns.barplot(data=df, x="alone", y="survived", hue="alone", errorbar=None, legend=False, ax=ax[1, 1])
ax[1, 1].set_title("Survival Rate: Alone vs With Family")
sns.boxplot(data=df, x="survived", y="fare", hue="survived", legend=False, ax=ax[1, 2])
ax[1, 2].set_ylim(0, 150)
ax[1, 2].set_title("Fare vs Survival")
plt.suptitle("Factors Affecting Survival", fontsize=16, fontweight="bold")
plt.tight_layout()
plt.savefig("output/2_survival_factors.png", dpi=150)
plt.show()

# Gender x Class combined
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="pclass", y="survived", hue="sex", errorbar=None)
plt.title("Survival Rate by Class and Gender")
plt.tight_layout()
plt.savefig("output/3_class_gender.png", dpi=150)
plt.show()

# 6. CORRELATION
corr = df.select_dtypes(include=np.number).corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig("output/4_correlation.png", dpi=150)
plt.show()

corr_surv = corr["survived"].drop("survived").sort_values(key=abs, ascending=False)
print("\nCorrelation with survival:\n", corr_surv.round(3))

# 7. KEY INSIGHTS
overall = df["survived"].mean()
by_sex = df.groupby("sex")["survived"].mean()
by_class = df.groupby("pclass")["survived"].mean()
by_age = df.groupby("age_group", observed=True)["survived"].mean()
by_alone = df.groupby("alone")["survived"].mean()

insights = [
    f"Overall survival rate was {overall:.1%}.",
    f"Women survived at {by_sex['female']:.1%} vs men at {by_sex['male']:.1%}. Gender was the strongest factor.",
    f"1st class survival was {by_class[1]:.1%}, while 3rd class was only {by_class[3]:.1%}.",
    f"Children had a survival rate of {by_age['Child']:.1%}, higher than other age groups.",
    f"Passengers travelling with family survived at {by_alone[False]:.1%} vs {by_alone[True]:.1%} for those alone.",
    f"Fare is positively related to survival (correlation = {corr.loc['fare', 'survived']:.2f}); "
    f"class is negatively related ({corr.loc['pclass', 'survived']:.2f}).",
]
print("\n===== KEY INSIGHTS =====")
for i, text in enumerate(insights, 1):
    print(f"{i}. {text}")

# 8. STRUCTURED REPORT
report = f"""# EDA Report: Titanic Dataset

## 1. Dataset Overview
- Rows: {df.shape[0]}, Columns: {df.shape[1]}
- Target variable: `survived`
- Cleaning: dropped `deck` (too many missing values), filled `age` with median, `embarked` with mode

## 2. Key Findings
""" + "\n".join(f"{i}. {t}" for i, t in enumerate(insights, 1)) + """

## 3. Conclusion
Gender, passenger class and age were the main factors influencing survival.
Women, children and first-class passengers had the highest chances of survival.
"""
with open("output/eda_report.md", "w") as f:
    f.write(report)
print("\nReport and charts saved in the 'output' folder.")
```
